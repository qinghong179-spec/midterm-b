````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# import sqlite3
# ==========================================
# (2) 建立資料庫與初始資料
# ==========================================
# 每次執行時先刪除舊資料表（確保乾淨），然後建立 menu 與 orders 資料表，並寫入初始資料。
import sqlite3

def init_db():
    conn = sqlite3.connect('coffee_shop.db')
    cursor = conn.cursor()
    try:
        # 每次執行先刪除舊資料表，確保環境重置
        cursor.execute("DROP TABLE IF EXISTS orders")
        cursor.execute("DROP TABLE IF EXISTS menu")

        # 建立菜單資料表 (menu)
        cursor.execute('''
            CREATE TABLE menu (
                item_name TEXT PRIMARY KEY,
                price INTEGER,
                stock INTEGER
            )
        ''')

        # 建立訂單資料表 (orders)
        cursor.execute('''
            CREATE TABLE orders (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                name TEXT,
                item_name TEXT,
                status TEXT
            )
        ''')

        # 放入初始菜單資料
        initial_menu = [
            ('美式咖啡', 60, 10),
            ('拿鐵', 80, 8),
            ('卡布奇諾', 85, 5),
            ('抹茶拿鐵', 90, 6)
        ]
        cursor.executemany("INSERT INTO menu (item_name, price, stock) VALUES (?, ?, ?)", initial_menu)

        conn.commit()
        print("資料庫建立成功，初始資料已載入！")
    finally:
        conn.close()

# 執行初始化函式
init_db()
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
# ==========================================
# (3) 顯示資料表
# ==========================================
# 使用 pandas 讀取並顯示目前資料庫中的 menu 與 orders 內容
import sqlite3
import pandas as pd
from IPython.display import display

def show_tables():
    conn = sqlite3.connect('coffee_shop.db')
    try:
        print("【 菜單資料表 (menu) 】")
        df_menu = pd.read_sql_query("SELECT * FROM menu", conn)
        display(df_menu)

        print("\n【 訂單資料表 (orders) 】")
        df_orders = pd.read_sql_query("SELECT * FROM orders", conn)
        display(df_orders)
    finally:
        conn.close()

# 執行顯示資料表函式
show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 品項, 數量, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
# ==========================================
# (4) AI 解讀函式
# ==========================================
# 透過 Groq API 解析使用者的一句話，轉成「動作,品項,數量」格式，並具備快取與 429 重試機制
# ==========================================
# (4) AI 解讀函式
# ==========================================
# 透過 Groq API 解析使用者的一句話，轉成「動作,品項,數量」格式，並具備快取與 429 重試機制
import re
import time

# 建立快取字典，避免同一句話重複呼叫 AI
ai_cache = {}

def parse_with_ai(user_input):
    # 若該句話已經查詢過，直接從快取回傳
    if user_input in ai_cache:
        return ai_cache[user_input]

    prompt = f"""
請分析以下使用者語句，並嚴格以「動作,品項,數量」的格式回傳，不要有其他贅字。
- 動作只能是：訂購、取消、查詢
- 品項只能是：美式咖啡、拿鐵、卡布奇諾、抹茶拿鐵
- 如果使用者沒有說數量，預設為 1
- 如果無法判斷或不屬於上述內容，動作請填「未知」、品項填「無」、數量填「1」

使用者說的話：「{user_input}」
"""

    max_retries = 3
    response_text = "未知,無,1"

    for attempt in range(max_retries):
        try:
            chat_completion = client.chat.completions.create(
                model=MODEL,
                messages=[{"role": "user", "content": prompt}],
                temperature=0,
                reasoning_effort=REASONING_EFFORT
            )
            response_text = chat_completion.choices[0].message.content.strip()
            time.sleep(2)  # 每次呼叫後暫停 2 秒
            break
        except Exception as e:
            # 遇到 429 超過使用量錯誤時，等待 30 秒重試
            if "429" in str(e) or "rate_limit" in str(e).lower():
                print(f"觸發 API 頻率限制 (429)，等待 30 秒後重試 ({attempt+1}/{max_retries})...")
                time.sleep(30)
            else:
                print(f"呼叫 AI 發生錯誤: {e}")
                break

    # 嚴格驗證並清洗 AI 回覆內容（防呆機制）
    parts = response_text.split(',')
    if len(parts) >= 3:
        raw_action = parts[0].strip()
        raw_item = parts[1].strip()
        raw_qty = parts[2].strip()
    else:
        raw_action, raw_item, raw_qty = "未知", "無", "1"

    # 動作白名單檢查
    action = raw_action if raw_action in ["訂購", "取消", "查詢"] else "未知"

    # 品項白名單檢查
    valid_items = ["美式咖啡", "拿鐵", "卡布奇諾", "抹茶拿鐵"]
    item = raw_item if raw_item in valid_items else "無"

    # 數量正規表示式提取與檢查
    qty_match = re.search(r'\d+', raw_qty)
    if qty_match:
        qty = int(qty_match.group())
    else:
        qty = 1

    result = (action, item, qty)
    ai_cache[user_input] = result
    return result
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
# ==========================================
# (5) AI 解讀測試
# ==========================================
# 對 10 個測試案例進行 AI 解讀測試，並印出原始回覆、系統解讀與通過幾句
test_cases = [
    (1, "正常", "我想訂購一杯拿鐵", "訂購", "拿鐵", 1),
    (2, "正常", "幫我買兩杯美式咖啡", "訂購", "美式咖啡", 2),
    (3, "正常", "查一下卡布奇諾還有幾杯", "查詢", "卡布奇諾", 1),
    (4, "正常", "取消拿鐵", "取消", "拿鐵", 1),
    (5, "正常", "我要訂購三杯抹茶拿鐵", "訂購", "抹茶拿鐵", 3),
    (6, "錯誤", "今天天氣真好", "未知", "無", 1),
    (7, "錯誤", "隨便給我一杯飲料", "未知", "無", 1),
    (8, "邊界", "我想訂購四杯美式咖啡", "訂購", "美式咖啡", 4),
    (9, "正常", "查詢拿鐵的庫存", "查詢", "拿鐵", 1),
    (10, "正常", "取消美式咖啡", "取消", "美式咖啡", 1)
]

passed_count = 0

print("=== 開始 AI 解讀測試 ===")
for case_id, t_type, text, exp_action, exp_item, exp_qty in test_cases:
    # 取得系統解讀結果
    act, itm, qty = parse_with_ai(text)

    # 檢查是否符合預期
    is_passed = (act == exp_action) and (itm == exp_item) and (qty == exp_qty)
    if is_passed:
        passed_count += 1
        status_str = "【通過】"
    else:
        status_str = "【失敗】"

    print(f"測試 #{case_id} ({t_type}) | 語句: 「{text}」")
    print(f"   -> 系統解讀: 動作={act}, 品項={itm}, 數量={qty}")
    print(f"   -> 預期結果: 動作={exp_action}, 品項={exp_item}, 數量={exp_qty} => {status_str}\n")

print(f"測試完成！總共通過 {passed_count} / {len(test_cases)} 句。")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
# ==========================================
# (6) 功能函式
# ==========================================
# 處理訂購、取消與查詢的核心商業邏輯，包含例外處理與資料庫異動
import sqlite3

def process_order(name, user_input):
    # 1. 透過 AI 解析語意
    action, item_name, qty = parse_with_ai(user_input)

    # 若動作無法辨識
    if action == "未知" or item_name == "無":
        return "無法判斷您的需求。目前系統提供的品項有：美式咖啡、拿鐵、卡布奇諾、抹茶拿鐵。"

    conn = sqlite3.connect('coffee_shop.db')
    cursor = conn.cursor()
    try:
        # --- 功能：查詢 ---
        if action == "查詢":
            cursor.execute("SELECT stock FROM menu WHERE item_name = ?", (item_name,))
            row = cursor.fetchone()
            if row is None:
                return f"找不到品項「{item_name}」。"
            return f"「{item_name}」目前的剩餘庫存為 {row[0]} 杯。"

        # --- 檢查：訂購與取消需要填寫姓名 ---
        if not name or name.strip() == "":
            return "錯誤：進行「訂購」或「取消」操作時，必須填寫姓名！"

        # --- 功能：訂購 ---
        if action == "訂購":
            # 檢查數量是否在 1~3 範圍內
            if qty < 1 or qty > 3:
                return f"拒絕訂購：數量 {qty} 不符合規定（每次訂購數量限制為 1 至 3 杯）。"

            # 檢查品項與庫存
            cursor.execute("SELECT price, stock FROM menu WHERE item_name = ?", (item_name,))
            row = cursor.fetchone()
            if row is None:
                return f"找不到品項「{item_name}」。"

            price, stock = row[0], row[1]

            # 庫存不足例外檢查
            if stock < qty:
                return f"訂購失敗：「{item_name}」庫存不足（剩餘 {stock} 杯，欲訂購 {qty} 杯）。"

            # 執行資料庫更新：扣除庫存、新增訂單紀錄
            new_stock = stock - qty
            cursor.execute("UPDATE menu SET stock = ? WHERE item_name = ?", (new_stock, item_name))
            cursor.execute("INSERT INTO orders (name, item_name, status) VALUES (?, ?, ?)", (name, item_name, "訂購成功"))
            conn.commit()

            total_price = price * qty
            return f"訂購成功！{name} 訂購了 {qty} 杯 {item_name}，應付總金額 {total_price} 元。目前庫存剩餘 {new_stock} 杯。"

        # --- 功能：取消 ---
        elif action == "取消":
            # 尋找該學生對該品項的最新一筆有效訂單
            cursor.execute("SELECT id FROM orders WHERE name = ? AND item_name = ? AND status = '訂購成功' ORDER BY id DESC LIMIT 1", (name, item_name))
            row = cursor.fetchone()

            # 查無訂單例外檢查
            if row is None:
                return f"取消失敗：查無 {name} 訂購「{item_name}」的有效紀錄。"

            order_id = row[0]

            # 更新訂單狀態為已取消
            cursor.execute("UPDATE orders SET status = '已取消' WHERE id = ?", (order_id,))

            # 恢復該品項庫存 (預設取消 1 杯)
            cursor.execute("UPDATE menu SET stock = stock + 1 WHERE item_name = ?", (item_name,))
            conn.commit()

            # 取得更新後的庫存
            cursor.execute("SELECT stock FROM menu WHERE item_name = ?", (item_name,))
            stock_row = cursor.fetchone()
            current_stock = stock_row[0] if stock_row else 0

            return f"取消成功：已將 {name} 的「{item_name}」訂單改為已取消，庫存已釋出（目前剩餘 {current_stock} 杯）。"

        return "未知的操作指令。"
    finally:
        conn.close()
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
# ==========================================
# (7) 主流程與流程測試
# ==========================================
# 依序執行 8 個步驟的流程測試，驗證系統整合結果
import sqlite3

flow_tests = [
    (1, "王小明", "我要訂購美式咖啡"),
    (2, "", "美式咖啡剩幾杯"),
    (3, "李大華", "我要訂購不存在的飲料"),
    (4, "", "我要訂購拿鐵"),
    (5, "張美美", "幫我點拿鐵"),
    (6, "", "哈囉你好"),
    (7, "", "查一下美式咖啡還有多少"),
    (8, "王小明", "我要訂購美式咖啡")
]

print("=== 開始主流程與流程測試 ===")
for step_id, test_name, test_input in flow_tests:
    # 每次測試前重新初始化資料庫，確保測試環境乾淨獨立
    init_db()

    # 針對第 1、8 步先讓王小明買一次，製造初始情境以符合預期
    if step_id in [1, 8]:
        process_order("王小明", "我要訂購美式咖啡")

    result_msg = process_order(test_name, test_input)
    print(f"流程步驟 #{step_id} | 姓名: {test_name or '(免填)'} | 輸入: 「{test_input}」")
    print(f"   -> 系統回應: {result_msg}\n")
print("=== 流程測試完畢 ===")
````
